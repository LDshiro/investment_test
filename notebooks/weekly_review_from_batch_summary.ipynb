# Weekly review from `batch_summary.csv`

This notebook reads a historical shadow batch summary, enriches it with packet metadata when available, and writes a weekly review bundle under `reports/weekly_review_latest/`.

```python
from pathlib import Path
import sys

REPO_ROOT = Path.cwd()
SRC_DIR = REPO_ROOT / 'src'
if str(SRC_DIR) not in sys.path:
    sys.path.insert(0, str(SRC_DIR))

from leadlag.reporting.weekly_review import load_batch_summary, enrich_batch_summary, aggregate_weekly, generate_weekly_review
```

```python
# Point this either at a batch directory or directly at batch_summary.csv
BATCH_SUMMARY_PATH = Path('runs/your_batch_dir/batch_summary.csv')
OUTPUT_DIR = Path('reports/weekly_review_latest')
```

```python
daily_df = enrich_batch_summary(load_batch_summary(batch_summary_path=BATCH_SUMMARY_PATH))
weekly_df = aggregate_weekly(daily_df)
daily_df.head()
```
